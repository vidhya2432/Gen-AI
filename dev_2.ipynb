```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

x=np.linspace(0,10,100)
y=np.sin(x)


plt.plot(x,y)
plt.title("Sin Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)

plt.show()
```

```python
x=np.linspace(0,10,100)
y=np.cos(x)


plt.plot(x,y)
plt.title("Cos Wave")
plt.xlabel("x")
plt.ylabel("Cos(x)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(0,10,100)
fig,ax=plt.subplots(2,1 , figsize=(6,6))
ax[0].plot(x,np.sin(x))
ax[0].set_title("Sin Wave")
ax[0].set_xlabel("x")
ax[0].set_ylabel("sin(x)")
ax[0].grid(True)
ax[1].plot(x,np.cos(x))
ax[1].set_title("Cos Wave")
ax[1].set_xlabel("x")
ax[1].set_ylabel("cos(x)")
ax[1].grid(True)


plt.tight_layout()
plt.show()
```
