```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x, y):
    return np.sin(x) * np.cos(y)

# Create x and y values
x = np.linspace(0, 5, 50)
y = np.linspace(0, 5, 40)

# Create grid
X, Y = np.meshgrid(x, y)

# Calculate Z values
Z = f(X, Y)

# Contour plot
plt.contour(X, Y, Z, colors='black')

# Filled contour plot
plt.contourf(X, Y, Z, 20, cmap='RdGy')
plt.colorbar()

# Image plot
plt.imshow(
    Z,
    extent=[0, 5, 0, 5],
    origin='lower',
    cmap='RdGy',
    alpha=0.5
)

plt.colorbar()

# Set equal aspect ratio
plt.axis('image')

# Add contour lines
contours = plt.contour(X, Y, Z, 3, colors='black')

# Add labels to contour lines
plt.clabel(contours, inline=True, fontsize=8)

# Labels and title
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Contour and Image Plot")

# Display the plot
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
# Load CSV file
tech_stocks_data = pd.read_csv("GOOG_MSFT_FB_AMZN_data.csv")
# Sort data by date
tech_stocks_data.sort_values("Period", ascending=True, inplace=True)
# Extract stock prices
google = tech_stocks_data["Alphabet Inc Price"]
amazon = tech_stocks_data["Amazon.com Inc Price"]
facebook = tech_stocks_data["Facebook Inc Price"]
microsoft = tech_stocks_data["Microsoft Corp Price"]
# Extract dates
dates = tech_stocks_data["Period"]
# Extract years
x = []
for date in tech_stocks_data["Period"]:
    x.append(
        datetime.strptime(
            date, "%Y-%m-%d %H:%M:%S"
        ).year
    )
# Create figure
plt.figure(figsize=(16, 12))
# Google
plt.subplot(2, 2, 1)
plt.plot(dates, google)
plt.title("Alphabet (GOOG) (GOOGL) Stock Price")
# Amazon
plt.subplot(2, 2, 2)
plt.plot(dates, amazon)
plt.title("Amazon (AMZN) Stock Price")
# Facebook
plt.subplot(2, 2, 3)
plt.plot(dates, facebook)
plt.title("Facebook (FB) Stock Price")
# Microsoft
plt.subplot(2, 2, 4)
plt.plot(dates, microsoft)
plt.title("Microsoft (MSFT) Stock Price")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
plt.plot(x, y, marker='o')
plt.title("Sales Performance")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.text(4, 25, "Sales Point")
plt.annotate("Highest Sales", xy=(5, 30),xytext=(3.5, 35), arrowprops=dict(arrowstyle="->"))
plt.grid(True)

plt.show()
```
